```python
# 50.Perform the complete data-cleaning process from loading the dataset to filling missing values using the median.

import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("diabetes.csv")

# Columns where 0 is invalid
zero_columns = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]

# Replace 0 with NaN
df[zero_columns] = df[zero_columns].replace(0, np.nan)

# Fill missing values using median
df[zero_columns] = df[zero_columns].fillna(df[zero_columns].median())

# Check missing values
print(df.isnull().sum())

# Check information
df.info()

# Display statistical summary
print(df.describe())
```
