```python
#Functions
def calculateAverage(a, b):
    return (a + b) / 2

result = calculateAverage(10, 5);
print(result);

def calculateSummation(a, b):
    return (a + b)

sum = calculateSummation(10, 10);
print(sum);

def calculateSubstraction(a, b):
    return (a - b)

sub = calculateSubstraction(50, 40);
print(sub);

def calculateMultiplication(a, b):
    return (a * b)

mul = calculateMultiplication(10, 10);
print(mul);

def calculateDivision(a, b):
    return (a / b)

div = calculateDivision(40, 40);
print(div);
```

```python
#pandas for load csv files
import pandas as pd

#df - data frames
df = pd.read_csv("Data1.csv")\

#first 5 records
df.head()

#range of data
df.tail()

#No of rows & cols
df.shape

#print col names
df.columns

#data types
df.dtypes

#general informations
df.info()

#unique values
df["Gender"].unique()

#converting into ctegorical variables
df["Gender"].value_counts()

#statical analysis
df.describe()

#mean value for study hours
r1 = df["Study_Hours"].mean()
print("Mean: ", r1);

r2 = df["Study_Hours"].min()
print("Minimum: ", r2)

r3 = df["Study_Hours"].std()
print("Standard deviation: ", r3)

# No of missing values
df.isnull().sum()


# total no of mssing values in the file
df.isnull().sum().sum()
```
